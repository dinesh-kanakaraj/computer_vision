```python
import cv2
from insightface.app import FaceAnalysis
```

```python



app = FaceAnalysis(
    name="buffalo_l",
    providers=["CPUExecutionProvider"]
)

app.prepare(
    ctx_id=-1,
    det_size=(640, 640)
)


image = cv2.imread(r"D:\Metamorphosis\facial_recognition\train\255c5077cbd378cee72b1cddcc90520c.jpg")

faces = app.get(image)

print("Faces detected:", len(faces))


for face in faces:

    print("Bounding box:")
    print(face.bbox)

    print("\nEmbedding shape:")
    print(face.normed_embedding.shape)

    print("\nFirst 10 embedding values:")
    print(face.normed_embedding[:10])
```

```python

import numpy as np



def get_embedding(image_path):
    image = cv2.imread(image_path)

    if image is None:
        raise FileNotFoundError(f"Could not load image: {image_path}")

    faces = app.get(image)

    if len(faces) == 0:
        raise ValueError(f"No face detected in: {image_path}")

    if len(faces) > 1:
        print(f"Warning: {len(faces)} faces detected. Using the first face.")

    return faces[0].normed_embedding


def compare_faces(image1_path, image2_path):
    embedding1 = get_embedding(image1_path)
    embedding2 = get_embedding(image2_path)

    similarity = np.dot(embedding1, embedding2)

    return float(similarity)


image1 = r"D:\Metamorphosis\facial_recognition\train\255c5077cbd378cee72b1cddcc90520c.jpg"
image2 = r"d:\Metamorphosis\facial_recognition\test\images.jfif"

score = compare_faces(image1, image2)

print("Similarity score:", score)
```
